**Year / Timeline:** Data collected during 2017 to 2024
* Date - Represents the date on which power generation was recorded.
* Region - Represents the geographical region associated with the power generation data.
* State Name - Represents the Indian state associated with the power generation record.
* Gen Actual (MU) - Represents the actual electricity generated during the day, measured in Million Units (MU).
* Gen Programmed (MU) - Represents the planned or programmed electricity generation for the day, measured in Million Units (MU).
* Capacity (MW) - Represents the power generation capacity, measured in Megawatts (MW).
* Wind Energy (MU) - Represents the electricity generated from wind energy, measured in Million Units (MU).
* Solar Energy (MU) - Represents the electricity generated from solar energy, measured in Million Units (MU).
* Other Renewable Energy (MU) - Represents the electricity generated from other renewable energy sources measured in Million Units (MU).
* Total Renewable Energy (MU) - Represents the combined electricity generation from wind, solar, and other renewable energy sources, measured in Million Units (MU).
* Total Gen With RE (MU) - Represents the total electricity generation including renewable energy generation, measured in Million Units (MU).

* It shows which regions have the most data registered and which one is the least.

* From this The Northern region has the highest number of records (above 15,000), followed by North Eastern, Southern, Eastern, and Western (around 8,500).

**columns used:**

 1.Region- Y axis -shows the available region

 2.Number of Records -x axis - shows count of records

Insights:

* Data collection is highest in the Northern region and lowest in the Western region.
* The dataset is slightly imbalanced, meaning some regions have collected more operational data points than others.

```python
#SCATTER PLOT(CAPACITY VS TOTAL RENEWABLE ENERGY)

plt.figure(figsize=(9, 6))
sns.scatterplot( data=df, x='capacity', y='total_renewable_energy',hue='region')

plt.title('Capacity vs Total Renewable Energy')
plt.xlabel('Capacity')
plt.ylabel('Total Renewable Energy')
plt.show()
```

## Interpretation of your above chart

 * This is a scatterplot that shows a relationship between capacity and the total renewable energy
 * the purpose to create this to understand whether areas with higher power-generation capacity also tend to produce more renewable energy.

* The chart shows that renewable energy generation generally increases as capacity increases, and the relationship is not exactly same for every observation.

 * Western (Purple) and Southern (Red) regions achieve some of the highest total renewable energy values, peaking near 270 units respectively.
 * Northern (Green) also shows strong generation performance, reaching up to 230 units.
 * Eastern (Blue) and North Eastern (Orange) regions stay clustered near the bottom (close to 0),with respect to  capacity.

**columns used:**

 1.Capacity - X-axis	- Shows the available generation capacity

 2.Total Renewable Energy - Y-axis -	Shows renewable energy generation

 3.Region - Hue - Compares the pattern across regions

 Insights:

* From this chart we come to know that capacity alone does not guarantee high renewable energy production. Regional geography plays a crucial role in how efficiently capacity translates into actual energy output

```python
month_energy = ( df.groupby('Month')[['solar_energy','wind_energy','other_renewable_energy','total_renewable_energy']].mean().reset_index())
month_energy
```

```python
# Monthly comparison -average of solar,wind,other renewable energy and the total renewable energy
plt.figure(figsize=(8, 4))

sns.lineplot(
    data=month_energy,
    x='Month',
    y='solar_energy',
    marker='o',
    label='Solar Energy')

sns.lineplot(
    data=month_energy,
    x='Month',
    y='wind_energy',
    marker='o',
    label='Wind Energy')

sns.lineplot(
    data=month_energy,
    x='Month',
    y='other_renewable_energy',
    marker='o',
    label='other Renewable Energy')
sns.lineplot(
    data=month_energy,
    x='Month',
    y='total_renewable_energy',
    marker='o',
    label='total Renewable Energy')


plt.title('Average Renewable Energy by Month')
plt.xlabel('Month')
plt.ylabel('Average Energy')
plt.xticks(range(1, 13))
plt.legend()
plt.grid(alpha=0.3)

plt.show()
```

## Interpretation of your above chart

* This is a multi-line graph comparing the average monthly energy generation for different renewable energy sources across the 12 months of the year (January to December).

* This chart shows how energy production changes seasonally for each source:

1.Solar Energy (Blue Line): Peaks during spring and early summer (Months 2–5) around 9 units, drops during the monsoon season (Months 6–8) to around 7 units, and slightly recovers in early autumn before falling again.

2.Wind Energy (Orange Line): Remains low early in the year (around 3.5–4.5 units), surges during the monsoon and summer months (Months 6–8) reaching a peak near 11 units in Month 7, and falls sharply back down by Month 10.

3.Other Renewable Energy (Green Line): Remains completely flat and stable throughout the entire year at around 1 unit.

4.Total Renewable Energy (Red Line): Combines all sources and reaches its highest peak mid-year (Months 5–8) around 18–19 units.

 * how solar and wind power complement each other throughout the year, keeping total renewable energy output high during the middle months of the year.

**columns used:**

1.Month -x-axis -  1 to 12 month

2.Average Energy - y-axis

3.Energy source types - legend -(Solar Energy, Wind Energy, other Renewable Energy, total Renewable Energy)




```python
region_year = (
    df.groupby(['Year', 'region'])['total_renewable_energy'].sum() .reset_index())

plt.figure(figsize=(8, 4))

sns.lineplot(
    data=region_year,
    x='Year',
    y='total_renewable_energy',
    hue='region',
    marker='o'
)

plt.title('Yearly Renewable Energy Trend by Region')
plt.xlabel('Year')
plt.ylabel('Total Renewable Energy')
plt.legend(title='Region')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## Interpretation of your above chart

* A multi-line chart tracking total renewable energy production over years (2020–2024), separated by region.

* It shows which regions are driving overall energy growth over time.
* Southern and Western regions generate the highest total energy, showing  upward growth from 2020 to 2023.

* Northern region shows moderate growth.

* Eastern and North Eastern regions remain near zero throughout all five years.

**Columns used:**

1.Year - X-axis - yearly time period

2.Total Renewable Energy - Value - Y-axis

3.Region - Category - Legend

**Insights:**

* India's renewable energy growth is almost entirely driven by the Southern, Western, and Northern regions.
